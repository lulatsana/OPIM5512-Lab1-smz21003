````python
import pandas as pd
import matplotlib.pyplot as plt

# The cleaned demand file (all of New England - same for both campuses). Same file sits in YOUR repo
# at data/clean/ — this reads the class copy so there's nothing to configure tonight.
url = "https://raw.githubusercontent.com/drdave-teaching/opim5512-lab1-template/main/data/clean/demand_hourly.csv"
dem = pd.read_csv(url, parse_dates=["hour"])

print(len(dem), "hours |", dem["hour"].min(), "->", dem["hour"].max())
dem.head()

CAMPUS = "hartford"
````

### How this was cleaned (read it — don't code it tonight)

The raw file was an **ISO New England** report where every line starts with a tag: `C` comment,
`H` header (two of them), `D` data, and a `T` summary footer. The cleaning kept the `D` rows and
turned **Hour Ending 1–24** into a timestamp for the hour that **begins** (Hour Ending 01 = the
00:00–01:00 hour). Result: one row per hour.

| column | meaning | units |
|---|---|---|
| `hour` | the hour that **begins** at this timestamp | — |
| `load_mw` | total New England electricity demand | MW |

*(Want the full version where you do the cleaning yourself? It's the "extended" Lab 1 in the class repo.)*


## 1. The line plot (already written — just run it)

Demand across the month. **Plot against time, not row number** — the question is *when*.

````python
ax = dem.plot(x="hour", y="load_mw", figsize=(12, 4), legend=False,
              title="Hourly electricity demand, New England, Jul 20 - Aug 19 2026")
ax.set_xlabel("date"); ax.set_ylabel("demand (MW)")
ax.get_figure().savefig("demand_line.png", dpi=150, bbox_inches="tight")
print("saved demand_line.png")
````

## 2. Your turn: one histogram
